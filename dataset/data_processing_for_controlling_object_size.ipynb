```python
import json
import os
import xml.etree.ElementTree as ET
from PIL import Image
import random
from tqdm import tqdm
```

```python
dataset_path = "F:\\pest_data\\Multitask_or_multimodality\\annotated_data"
```

```python
def check_annotation(
        xml_file:str, 
        image_path:str):
    xml_tree = ET.parse(xml_file)
    root = xml_tree.getroot()
    width = int(root.find("size").find("width").text)
    height = int(root.find("size").find("height").text)
    image = Image.open(image_path)
    if (image.size[0] != width) or (image.size[1]!=height):
        print(f"Image size mismatch: {image_path} - {image.size} vs {width}x{height}")

    file_name = root.find("filename").text

    if file_name != os.path.basename(image_path):
        print(f"Correct file name mismatch: {image_path} - {file_name}")
        file_name = os.path.basename(image_path)
        root.find("filename").text = file_name
        xml_tree.write(xml_file)

    for obj in root.findall("object"):
        name = obj.find("name").text
        bnding_box = obj.find("bndbox")
        xmin = int(bnding_box.find("xmin").text)
        ymin = int(bnding_box.find("ymin").text)
        xmax = int(bnding_box.find("xmax").text)
        ymax = int(bnding_box.find("ymax").text)
        if xmin < 0 or ymin < 0 or xmax > width or ymax > height:
            print(f"Bounding box out of image bounds: {image_path} - {name} ({xmin},{ymin},{xmax},{ymax})")
            # bnding_box.find("xmin").text = str(max(0, xmin))
            # bnding_box.find("ymin").text = str(max(0, ymin))
            # bnding_box.find("xmax").text = str(min(width, xmax))
            # bnding_box.find("ymax").text = str(min(height, ymax))
            # xml_tree.write(xml_file)

        obj_width = xmax - xmin
        obj_height = ymax - ymin
        objects_size = obj_width * obj_height
        small_objects_threshold = width * height /100
        if objects_size <= small_objects_threshold:
            print(f"Small objects: {image_path} - {name} ({xmin},{ymin},{xmax},{ymax})")
    
```

```python
image_list = [file for file in os.listdir(dataset_path) if not file.endswith(".xml")]

for file in tqdm(os.listdir(dataset_path)):
    if file.endswith(".xml"):
        xml_file = os.path.join(dataset_path, file)
        image_path = os.path.join(dataset_path, file.replace(".xml", ".jpg"))

        check_annotation(xml_file, image_path)
```

```python
def check_yolo_label(txt_path):
    with open(txt_path, "r") as f:
        lines = f.readlines()
    for line in lines:
        parts = line.strip().split()
        if len(parts) != 5:
            continue
        class_id, x_center, y_center, width, height = map(float, parts)

        x_min = x_center - width / 2
        y_min = y_center - height / 2
        x_max = x_center + width / 2
        y_max = y_center + height / 2
        if x_min < 0 or y_min < 0 or x_max > 1 or y_max > 1:
            print(f"YOLO label out of bounds: {txt_path} - {line.strip()}")
            continue
```

```python
yolo_dataset_path = "F:\\pest_data\\Multitask_or_multimodality\\YOLO_06JAN25_ALL_INSECT"


train_label_path = os.path.join(yolo_dataset_path, "labels", "train")
train_image_path = os.path.join(yolo_dataset_path, "images", "train")
val_label_path = os.path.join(yolo_dataset_path, "labels", "val")
val_image_path = os.path.join(yolo_dataset_path, "images", "val")

for file in tqdm(os.listdir(train_label_path)):
    if file.endswith(".txt"):
        txt_path = os.path.join(train_label_path, file)
        check_yolo_label(txt_path)
for file in tqdm(os.listdir(val_label_path)):
    if file.endswith(".txt"):
        txt_path = os.path.join(val_label_path, file)
        check_yolo_label(txt_path)
```
