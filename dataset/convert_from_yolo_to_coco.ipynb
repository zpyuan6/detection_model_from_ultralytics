```python
import json
```

```python
def convert_coco_categoris(category_map, annotation_file_path):
    with open(annotation_file_path, 'r') as f:
        coco = json.load(f)


    categories = coco['categories']

    old_image_list = coco['annotations']

    id_map = {}

    for i, category in enumerate(categories):
        target_id = category_map[category['name'].upper()] + 1
        id_map[category['id']] = target_id

    new_image_list = []
    for image in old_image_list:
        i = image
        i["category_id"] = id_map[i["category_id"]]
        new_image_list.append(i)

    coco['annotations'] = new_image_list

    with open("X:\\pervasive_group\\PestProject\\COCO_07JAN25_ALL_INSECT\\coco_pest_2025_01_09\\annotations\\updatedselectedVal.json", 'w') as f:
        json.dump(coco, f)
    
```
